# YPNUS.com integration (Claude Cookbooks) — Anthropic client example

This notebook generates marketing content with Claude using the repository's Anthropic pattern and shows how to push it to YPNUS.com. Fill YPNUS API fields per your docs and set ANTHROPIC_API_KEY and YPNUS_API_KEY in a .env file.

```python
import os
from dotenv import load_dotenv
import requests

load_dotenv()
ANTHROPIC_API_KEY = os.getenv('ANTHROPIC_API_KEY')
YPNUS_API_KEY = os.getenv('YPNUS_API_KEY')

assert ANTHROPIC_API_KEY, 'Set ANTHROPIC_API_KEY in .env'
assert YPNUS_API_KEY, 'Set YPNUS_API_KEY in .env'
```

```python
from anthropic import Anthropic

def generate_marketing_content(prompt: str, model: str = 'claude-sonnet-4-6') -> str:
    """Generate marketing copy using the repo's Anthropic pattern.

    Returns the assistant text or raises on error.
    """
    client = Anthropic(api_key=ANTHROPIC_API_KEY)
    try:
        response = client.messages.create(
            model=model,
            max_tokens=800,
            temperature=0.7,
            messages=[{'role': 'user', 'content': prompt}],
        )
        return response.content[0].text.strip()
    except Exception as e:
        return f'<ERROR: {e}>'
```

```python
# Example: function to push content to YPNUS.com
# Update endpoint path and payload shape to match YPNUS API docs
YPNUS_API_BASE = 'https://api.ypnus.com/v1'  # replace with actual base URL from docs

def push_page_to_ypnus(title: str, body_html: str, zip_code: str = None) -> dict:
    url = f'{YPNUS_API_BASE}/pages'  # example endpoint — change as needed
    headers = {
        'Authorization': f'Bearer {YPNUS_API_KEY}',
        'Content-Type': 'application/json'
    }
    payload = {
        'title': title,
        'content_html': body_html,
        'territory_zip': zip_code,
    }
    resp = requests.post(url, json=payload, headers=headers)
    resp.raise_for_status()
    return resp.json()

# Example usage:
prompt = 'Write a concise, high-converting landing page for a mortgage broker serving 90210; include call-to-action and local SEO headings.'
generated = generate_marketing_content(prompt)
print('Generated preview:
', generated)
# response = push_page_to_ypnus('90210 Mortgage Landing', generated, zip_code='90210')
# print('YPNUS response:', response)
```
