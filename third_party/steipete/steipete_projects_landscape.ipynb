# Steipete Tooling Landscape: Skills-First Ingestion

This cookbook is optimized for immediate ingestion into Codex skills. It turns a curated steipete project catalog into lazy-loadable skill packs.

- Catalog file: `third_party/steipete/projects.yaml`
- Ingestion script: `third_party/steipete/ingest_projects_as_skills.py`
- Target output: `~/.codex/skills`


## Instant Ingestion

Run this once to generate an index skill plus one skill per project:

```bash
python3 third_party/steipete/ingest_projects_as_skills.py --dest ~/.codex/skills --overwrite
```

After generation, restart Codex so newly generated skills are discovered in-session.


````python
from pathlib import Path
import yaml

catalog_path = Path('third_party/steipete/projects.yaml')
payload = yaml.safe_load(catalog_path.read_text(encoding='utf-8'))
projects = payload['projects']

status_counts = {}
for project in projects:
    status = project.get('status', 'current')
    status_counts[status] = status_counts.get(status, 0) + 1

print(f"Catalog: {catalog_path}")
print(f"Projects: {len(projects)}")
print(f"Status breakdown: {status_counts}")
````

## Generated Skill Topology

The ingestion script creates:

- `steipete-tooling-index`: router skill for project discovery and branching
- `steipete-<project-slug>`: one focused skill per project

Each generated skill includes:

- `SKILL.md` with trigger-focused metadata
- `references/*.md` with project context
- `agents/openai.yaml` for UI discovery

This keeps metadata lightweight while allowing deep context only when a specific skill is selected.


## Recommended Usage Pattern

1. Load `steipete-tooling-index` when user asks broad ecosystem questions.
2. Branch to only the relevant `steipete-<project-slug>` skill(s).
3. Keep context lean by avoiding multi-skill fan-in unless needed.

This pattern maximizes speed and minimizes context pressure in long sessions.
