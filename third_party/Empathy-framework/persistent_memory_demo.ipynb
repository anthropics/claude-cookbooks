# Persistent Memory with Claude using Empathy Framework

This notebook demonstrates how to add persistent, cross-session memory to Claude using the [Empathy Framework](https://github.com/Smart-AI-Memory/empathy-framework).

**What you'll learn:**
- How to give Claude memory that persists across conversations
- How memory context improves response quality over time
- The five levels of AI-human collaboration

## Why Persistent Memory?

By default, each Claude conversation starts fresh. Persistent memory allows Claude to:
- Remember user preferences ("I prefer concise answers")
- Track ongoing projects across sessions
- Build context over time for more personalized assistance
- Anticipate needs based on historical patterns

```python
import os
from empathy_llm_toolkit import EmpathyLLM

# Set your API key (or use environment variable)
# os.environ["ANTHROPIC_API_KEY"] = "your-api-key-here"
```

## Basic Usage: Memory-Enhanced Conversations

The `EmpathyLLM` class wraps the Anthropic API and adds persistent memory storage.

```python
# Initialize with persistent memory
llm = EmpathyLLM(
    provider="anthropic",
    api_key=os.getenv("ANTHROPIC_API_KEY"),
    memory_enabled=True
)

# User ID identifies whose memory to use
USER_ID = "demo_user"
```

```python
# First interaction - establish a preference
response = await llm.interact(
    user_id=USER_ID,
    user_input="I'm a Python developer. I prefer concise code examples without excessive comments."
)
print(response.content)
```

```python
# Second interaction - Claude remembers the preference
response = await llm.interact(
    user_id=USER_ID,
    user_input="Show me how to read a JSON file"
)
print(response.content)

# Notice: Claude provides concise Python code without excessive comments,
# because it remembered the user's preference from the first interaction.
```

## Memory Persists Across Sessions

The key difference from regular Claude: memory survives when you close and reopen your application.

```python
# Simulate a new session (new LLM instance)
llm_new_session = EmpathyLLM(
    provider="anthropic",
    api_key=os.getenv("ANTHROPIC_API_KEY"),
    memory_enabled=True
)

# Same user, new session - memory is restored
response = await llm_new_session.interact(
    user_id=USER_ID,
    user_input="Show me how to write to a CSV file"
)
print(response.content)

# Claude still remembers: Python developer, prefers concise examples
```

## The Five Levels of Empathy

The Empathy Framework implements five levels of AI-human collaboration:

| Level | Name | Description |
|-------|------|-------------|
| 1 | Reactive | Standard request-response (baseline) |
| 2 | Informed | Uses stored context to personalize responses |
| 3 | Predictive | Anticipates needs based on patterns |
| 4 | Anticipatory | Proactively suggests before being asked |
| 5 | Collaborative | Full partnership with shared goals |

You can configure the empathy level:

```python
# Level 3: Predictive - anticipates based on patterns
response = await llm.interact(
    user_id=USER_ID,
    user_input="I'm starting a new data pipeline project",
    empathy_level=3
)
print(response.content)

# With Level 3, Claude might proactively suggest:
# - Libraries you've used before (pandas, etc.)
# - Patterns from your previous projects
# - Potential issues based on past experiences
```

## Viewing Stored Memory

You can inspect what the framework has learned about a user:

```python
# Get memory summary for user
memory_summary = await llm.get_memory_summary(user_id=USER_ID)
print("Stored preferences:")
for key, value in memory_summary.items():
    print(f"  {key}: {value}")
```

## Project Context

Memory can also be scoped to specific projects:

```python
# Different projects maintain separate context
response = await llm.interact(
    user_id=USER_ID,
    user_input="What's the status of this project?",
    context={
        "project": "data-pipeline-v2",
        "scope": "project"  # Use project-specific memory
    }
)
print(response.content)
```

## Privacy Controls

The framework includes built-in privacy features:

```python
# Clear all memory for a user
await llm.clear_memory(user_id=USER_ID)
print("Memory cleared for user")

# Or selectively forget specific information
await llm.forget(
    user_id=USER_ID,
    pattern="email"  # Forget anything containing 'email'
)
```
